Fit three nested regression models to `gtemp_land` (linear, quadratic, cubic). Compute AIC, AICc, and BIC for each using the Lecture 3 formulas:
$$\text{AIC}(k) = \log\hat{\sigma}_k^2 + \frac{n+2k}{n}, \quad \text{BIC}(k) = \log\hat{\sigma}_k^2 + \frac{k\log n}{n}$$
Which model does each criterion select?

```r
# 2(d): Model selection
# Helper function from Lecture 3
reg.stats <- function(fit) {
  n   <- length(fit$residuals)
  k   <- n - fit$df.residual       # number of parameters
  MSE <- sum(fit$residuals^2) / n
  AIC   <- log(MSE) + (n + 2*k) / n
  BIC   <- log(MSE) + k * log(n) / n
  AICc  <- log(MSE) + (n + k) / (n - k - 2)
  data.frame(k=k, MSE=MSE, AIC=AIC, AICc=AICc, BIC=BIC)
}

t_idx  <- time(gtemp_land)
t2_idx <- t_idx^2
t3_idx <- t_idx^3

# fit.1 <- lm(gtemp_land ~ ...)
# fit.2 <- lm(gtemp_land ~ ...)
# fit.3 <- lm(gtemp_land ~ ...)
# your code here
```
